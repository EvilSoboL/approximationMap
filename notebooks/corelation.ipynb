```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import r2_score
```

```python
coef_steam: list = [1.31, 1.11, 2.87, 2.22, 1.09]
viscosity_steam: list = [4.1, 69, 127.4, 6.3, 2.01]

def y1(x, a = 1.7, b = 2, c = -0.069):
    return a * (x - b) ** c

y_eq = [y1(i) for i in sorted(viscosity_steam)]

plt.plot(sorted(viscosity_steam), y_eq)
plt.scatter(viscosity_steam, coef_steam)

y_true = np.array(coef_steam)
y_pred = [y1(i) for i in viscosity_steam]

r2 = r2_score(y_true, y_pred)
print(r2)
```
