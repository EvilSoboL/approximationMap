```python
import numpy as np
import matplotlib.pyplot as plt

# Данные
x = np.linspace(0, 10, 10)
y = np.sin(x)
# Ошибка (например, стандартное отклонение или доверительный интервал)
yerr = 0.2

# Построение графика с доверительными интервалами в виде отрезков
plt.scatter(x, y)
plt.errorbar(x, y, yerr=yerr, fmt='o', capsize=5, label='Доверительный интервал')

# Настройки
plt.legend()
plt.show()
```
